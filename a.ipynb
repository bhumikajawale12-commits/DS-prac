```python
### 1. Import all the required Python Libraries
import pandas as pd
import numpy as np
```

```python
### 3. Load the Dataset into a Pandas DataFrame

# Download Dataset from kaggle
# df = pd.read_csv("titanic.csv")

url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
df = pd.read_csv(url)

df.head()
```

```python
### 4. Data Preprocessing

#finding missing values
print("Dimensions:", df.shape)
print("\nMissing values:\n", df.isnull().sum())
```

```python
# Replace missing Age values with mean age
df['Age'] = df['Age'].fillna(df['Age'].mean())
# Replace missing Embarked values with most frequent value
df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])
df.isnull().sum()
```

```python
# Replace missing cabin values with 'Unknown'
df['Cabin'] = df['Cabin'].fillna("Unknown")

# Display updated values
print("Updated Cabin column:")
print(df['Cabin'].head())

# Again Check number of missing values in Cabin column
print("Missing Cabin values:", df['Cabin'].isnull().sum())
df.isnull().sum()
```

```python
print("\nStatistics:")
display(df.describe(include="all"))
```

```python
### 5. Turn categorical variables into quantitative variables

from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

df['Sex'] = le.fit_transform(df['Sex'])
df['Embarked'] = le.fit_transform(df['Embarked'].astype(str))

display(df.head())
```

```python
### 5. Data Formatting and Data Normalization

from sklearn.preprocessing import MinMaxScaler

# Normalization
scaler = MinMaxScaler()

df[['Age', 'Fare']] = scaler.fit_transform(df[['Age', 'Fare']])

print(df[['Age', 'Fare']].head())

display(df.head())
```
